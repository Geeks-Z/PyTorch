## model.modules()

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(10, 20),
    nn.ReLU(),
    nn.Sequential(
        nn.Conv2d(1, 3, 3),
        nn.Sigmoid()
    )
)

for module in model.modules():
    print(module)
```

## model.named_modules()

```python
for name, module in model.named_modules():
    print(name, module)
```

## model.children()

```python
for child in model.children():
    print(child)
```

## model.named_children()

```python
for name, child in model.named_children():
    print(name, child)
```
